```python
import os
import re
import time
import numpy as np
import pandas as pd

import fitz
import docx
import faiss

from sentence_transformers import SentenceTransformer

print("✅ All required imports are working!")
```

```python
def load_txt(file_path):
    """
    Load text from a TXT file.
    """
    with open(file_path, "r", encoding="utf-8") as file:
        return file.read()
```

```python
def load_pdf(file_path):
    """
    Extract text from a PDF file using PyMuPDF.
    """
    document = fitz.open(file_path)

    text = ""

    for page in document:
        text += page.get_text()

    document.close()

    return text
```

```python
def load_docx(file_path):
    """
    Extract text from a DOCX file.
    """
    document = docx.Document(file_path)

    text = ""

    for paragraph in document.paragraphs:
        text += paragraph.text + "\n"

    return text
```

```python
def load_document(file_path):
    """
    Load a TXT, PDF, or DOCX document.
    """

    extension = os.path.splitext(file_path)[1].lower()

    if extension == ".txt":
        return load_txt(file_path)

    elif extension == ".pdf":
        return load_pdf(file_path)

    elif extension == ".docx":
        return load_docx(file_path)

    else:
        raise ValueError(
            f"Unsupported file format: {extension}"
        )
```

```python
file_path = "sample_documents/sample.txt"

text = load_document(file_path)

print(text)
```

```python
def clean_text(text):
    """
    Clean extracted document text.
    """

    # Replace tabs with spaces
    text = text.replace("\t", " ")

    # Replace multiple spaces with a single space
    text = re.sub(r"[ ]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    # Remove spaces at the beginning/end
    text = text.strip()

    return text
```

```python
cleaned_text = clean_text(text)

print(cleaned_text)
```

```python
print("Original characters :", len(text))
print("Cleaned characters  :", len(cleaned_text))
```

```python
def create_chunks(text, chunk_size=500, overlap=100):
    """
    Split text into overlapping chunks.
    """

    if overlap >= chunk_size:
        raise ValueError("Overlap must be smaller than chunk size.")

    chunks = []

    start = 0
    text_length = len(text)

    while start < text_length:

        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
chunks = create_chunks(cleaned_text)

print("Number of chunks:", len(chunks))
```

```python
from sentence_transformers import SentenceTransformer

model_name = "all-MiniLM-L6-v2"

model = SentenceTransformer(model_name)

print("✅ Embedding model loaded successfully!")
```

```python
print("Model:", model_name)
print("Embedding dimension:", model.get_sentence_embedding_dimension())
```

```python
embeddings = model.encode(
    chunks,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("✅ Embeddings generated!")
print("Shape:", embeddings.shape)
```

```python
print("First embedding:")
print(embeddings[0])
```

```python
test_sentences = [
    "Machine learning allows computers to learn from data.",
    "Artificial intelligence systems can learn patterns from information.",
    "I like eating pizza."
]

test_embeddings = model.encode(
    test_sentences,
    convert_to_numpy=True,
    normalize_embeddings=True
)

similarity_matrix = np.dot(test_embeddings, test_embeddings.T)

print("Similarity Matrix:")
print(similarity_matrix)
```

```python
# ==========================================
# Create FAISS Index
# ==========================================

embedding_dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(embedding_dimension)

index.add(embeddings.astype("float32"))

print("✅ FAISS index created successfully!")
print("Number of vectors:", index.ntotal)
print("Vector dimension:", embedding_dimension)
```

```python
print("FAISS index status")
print("-" * 40)
print("Total vectors:", index.ntotal)
print("Dimension:", index.d)
```

```python
query = "How does artificial intelligence learn from data?"

query_embedding = model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
)

scores, indices = index.search(
    query_embedding.astype("float32"),
    k=1
)

print("Query:")
print(query)

print("\nSimilarity score:")
print(scores[0][0])

print("\nRetrieved chunk:")
print(chunks[indices[0][0]])
```

```python
def process_documents(folder_path):
    """
    Load and process all supported documents in a folder.

    Returns:
        chunks: list of text chunks
        metadata: information about each chunk
    """

    all_chunks = []
    all_metadata = []

    supported_extensions = {".txt", ".pdf", ".docx"}

    for filename in sorted(os.listdir(folder_path)):

        file_path = os.path.join(folder_path, filename)

        if not os.path.isfile(file_path):
            continue

        extension = os.path.splitext(filename)[1].lower()

        if extension not in supported_extensions:
            continue

        # Load document
        document_text = load_document(file_path)

        # Clean text
        cleaned_document = clean_text(document_text)

        # Create chunks
        document_chunks = create_chunks(cleaned_document)

        # Store chunks and metadata
        for chunk_id, chunk in enumerate(document_chunks):

            all_chunks.append(chunk)

            all_metadata.append({
                "document": filename,
                "chunk_id": chunk_id
            })

    return all_chunks, all_metadata
```

```python
chunks, metadata = process_documents("sample_documents")

print("✅ Documents processed!")
print("Total chunks:", len(chunks))
print("Total metadata records:", len(metadata))
```

```python
for i in range(len(chunks)):

    print("=" * 60)
    print("Chunk:", i)
    print("Document:", metadata[i]["document"])
    print("Chunk ID:", metadata[i]["chunk_id"])
    print("Text:", chunks[i])
```

```python
embeddings = model.encode(
    chunks,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("✅ Embeddings generated!")
print("Shape:", embeddings.shape)
```

```python
embedding_dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(embedding_dimension)

index.add(
    embeddings.astype("float32")
)

print("✅ FAISS index rebuilt!")
print("Total vectors:", index.ntotal)
print("Vector dimension:", index.d)
```

```python
def semantic_search(query, top_k=3):
    """
    Search the FAISS index using a natural-language query.
    """

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores, indices = index.search(
        query_embedding.astype("float32"),
        top_k
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx == -1:
            continue

        results.append({
            "document": metadata[idx]["document"],
            "chunk_id": metadata[idx]["chunk_id"],
            "score": float(score),
            "text": chunks[idx]
        })

    return results
```

```python
results = semantic_search(
    "How do computers learn from data?",
    top_k=3
)

for result in results:

    print("=" * 70)
    print("📄 Document:", result["document"])
    print("📌 Chunk:", result["chunk_id"])
    print("📊 Similarity:", round(result["score"], 4))
    print("\n", result["text"])
```

```python
queries = [
    "What are neural networks?",
    "How can computers understand human language?",
    "How can database queries be made faster?",
    "What is the difference between AI and machine learning?",
    "What technology is commonly used for computer vision?"
]

for query in queries:

    print("\n" + "#" * 80)
    print("QUERY:", query)

    results = semantic_search(query, top_k=2)

    for result in results:

        print("\n📄", result["document"])
        print("📊 Score:", round(result["score"], 4))
        print("📝", result["text"][:300])
```
