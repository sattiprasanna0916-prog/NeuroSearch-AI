```python
import os
import fitz
import docx

def load_txt(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        return file.read()


def load_pdf(file_path):
    document = fitz.open(file_path)

    text = ""

    for page in document:
        text += page.get_text() + "\n"

    document.close()

    return text


def load_docx(file_path):
    document = docx.Document(file_path)

    text = ""

    for paragraph in document.paragraphs:
        text += paragraph.text + "\n"

    return text


def load_document(file_path):
    extension = os.path.splitext(file_path)[1].lower()

    if extension == ".txt":
        return load_txt(file_path)

    elif extension == ".pdf":
        return load_pdf(file_path)

    elif extension == ".docx":
        return load_docx(file_path)

    else:
        raise ValueError(
            f"Unsupported file format: {extension}"
        )


print("✅ TXT loader ready")
print("✅ PDF loader ready")
print("✅ DOCX loader ready")
print("✅ Multi-format document loader ready")
```

```python
import re

def clean_text(text):
    # Normalize line breaks
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove excessive spaces
    text = re.sub(r"[ \t]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    return text.strip()


print("✅ Text cleaning function ready!")
```

```python
def create_chunks(text, chunk_size=500, overlap=100):
    if not text:
        return []

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        # Move forward while keeping overlap
        start += chunk_size - overlap

    return chunks


print("✅ Chunking function ready!")
print("Chunk size: 500 characters")
print("Overlap: 100 characters")
```

```python
chunks = []
metadata = []

for filename in sorted(os.listdir(DOCUMENT_DIR)):

    file_path = os.path.join(DOCUMENT_DIR, filename)

    try:
        # Load document
        text = load_document(file_path)

        # Clean text
        cleaned_text = clean_text(text)

        # Create chunks
        document_chunks = create_chunks(cleaned_text)

        # Store chunks + metadata
        for chunk_id, chunk in enumerate(document_chunks):

            chunks.append(chunk)

            metadata.append({
                "document": filename,
                "chunk_id": chunk_id,
                "file_type": os.path.splitext(filename)[1].lower()
            })

        print(
            f"✅ {filename} → "
            f"{len(document_chunks)} chunks"
        )

    except Exception as e:
        print(f"❌ {filename} → {e}")


print("\n" + "=" * 50)
print("🎉 DOCUMENT INGESTION COMPLETE")
print("=" * 50)
print("Total chunks:", len(chunks))
print("Total metadata records:", len(metadata))
```

```python
import pickle
import os

chunks_path = os.path.join(INDEX_DIR, "chunks.pkl")
metadata_path = os.path.join(INDEX_DIR, "metadata.pkl")

with open(chunks_path, "wb") as f:
    pickle.dump(chunks, f)

with open(metadata_path, "wb") as f:
    pickle.dump(metadata, f)

print("✅ Chunks saved to Google Drive!")
print("✅ Metadata saved to Google Drive!")
print("\nSaved files:")
print("📦", chunks_path)
print("📦", metadata_path)
```
